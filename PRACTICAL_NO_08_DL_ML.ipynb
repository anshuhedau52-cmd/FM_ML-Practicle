```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering
from sklearn.decomposition import PCA

from scipy.cluster.hierarchy import dendrogram, linkage
```

```python
file = "wireless_signal_hierarchical_dataset.csv"
```

```python
data = pd.read_csv(file)
```

```python
if len(data.columns) == 1:
  print("Signal column detected. Fixing CSV format...")
  data = data.iloc[:,0].str.split(
      ",", expand=True
  )
```

```python
data.colomns = [
    "RSSI",
    "RSRP",
    "RSRQ",
    "Latency",
    "Packet_Loss"
]
```

```python
print("\nDataset Preview:")
print(data.head())

print("\nDataset Columns:")
print(data.columns)
```

```python
for col in data.columns:

    data[col] = pd.to_numeric(
        data[col],
        errors="coerce"
    )

data.dropna(inplace=True)

print("\nMissing Values:")
print(data.isnull().sum())
```

```python
X = data[
    [
        "RSSI",
        "RSRP",
        "RSRQ",
        "SNR",
        "Latency",
        "Packet_Loss"
    ]
]
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

linked = linkage(
    X_scaled,
    method="ward"
)

dendrogram(linked)

plt.title("Hierarchical Clustering Dendrogram")
plt.xlabel("Samples")
plt.ylabel("Distance")

plt.show()
```

```python
model = AgglomerativeClustering(
    n_clusters=3,
    linkage="ward"
)

data["Cluster"] = model.fit_predict(X_scaled)
print("\nCluster Count:")

print(
    data["Cluster"].value_counts()
)
```

```python
cluster_rssi = data.groupby(
    "Cluster"
)["RSSI"].mean()

print("\nAverage RSSI:")

print(cluster_rssi)
```

```python
sorted_clusters = cluster_rssi.sort_values(
    ascending=False
)

quality = {

    sorted_clusters.index[0]: "Good",

    sorted_clusters.index[1]: "Medium",

    sorted_clusters.index[2]: "Poor"
}

data["Signal_Quality"] = data["Cluster"].map(
    quality
)
```

```python
print("\nFinal Classification:")

print(
    data[
        [
            "RSSI",
            "RSRP",
            "SNR",
            "Cluster",
            "Signal_Quality"
        ]
    ]
)
```

```python
pca = PCA(
    n_components=2
)

X_pca = pca.fit_transform(
    X_scaled
)
```

```python
plt.figure(figsize=(8, 6))

sns.scatterplot(
    x=X_pca[:, 0],
    y=X_pca[:, 1],
    hue=data["Signal_Quality"],
    palette="Set1",
    s=100
)

plt.title(
    "Hierarchical Clustering of Wireless Signal Quality"
)

plt.xlabel(
    "Principal Component 1"
)

plt.ylabel(
    "Principal Component 2"
)

plt.show()
```
